# ZeeML Syntactic Burstiness, Yule's K & Readability Benchmark Analysis

This starter notebook evaluates the 600-passage full-text corpus accompanying the live interactive workbenches at **[ZeeML (`https://zeeml.com`)](https://zeeml.com)**:

- **AI Content & Burstiness Detector:** [https://zeeml.com/ai-detector](https://zeeml.com/ai-detector)

```python
import glob
import os
import pandas as pd

# Automatically locate the CSV inside /kaggle/input or current working directory
matches = glob.glob('/kaggle/input/**/zeeml_ai_cadence_entropy_corpus_v1.csv', recursive=True)
csv_path = matches[0] if matches else 'zeeml_ai_cadence_entropy_corpus_v1.csv'
print('Loading dataset from:', csv_path)

df = pd.read_csv(csv_path)
print(f'Dataset shape: {df.shape[0]} rows x {df.shape[1]} columns')
df.head(3)
```

```python
# Group-by summary across the 3 structural cadence cohorts
metrics = [
    'mean_sentence_words',
    'sentence_length_sd',
    'burstiness_cv',
    'shannon_lexical_entropy',
    'yules_k_diversity',
    'passive_voice_ratio',
    'flesch_kincaid_grade'
]
summary = df.groupby('cohort')[metrics].agg(['mean', 'std']).round(3)
summary
```

```python
# Verify Burstiness CV quantile separation across cohorts
for cohort, grp in df.groupby('cohort'):
    q25, med, q75 = grp['burstiness_cv'].quantile([0.25, 0.50, 0.75]).round(3)
    print(f'{cohort:20s} -> Median Burstiness CV: {med:.3f} (IQR: {q25:.3f} to {q75:.3f})')
```
